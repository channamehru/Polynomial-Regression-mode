# Polynomial Regression Model for Advertising Sales Prediction

## Project Objective
The goal of this project is to build a **Polynomial Regression model** to predict product sales based on advertising budgets for **TV**, **Radio**, and **Newspaper**.

Unlike simple linear regression, polynomial regression can capture **non-linear relationships** and show **diminishing returns** at higher spending levels.

## How This Project Helps Users
- Optimize advertising budgets for maximum ROI.
- Identify overspending and avoid diminishing returns.
- Allocate budgets effectively across multiple channels.
- Make data-driven marketing decisions quickly.
- Understand non-linear relationships in real-world sales data.


## Q1. Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

import warnings
warnings.filterwarnings('ignore')
```

## Q2. Load the Advertising Dataset

```python
# Load Advertising dataset from a public source
# You may also replace this with: df = pd.read_csv('Advertising.csv')
url = 'https://raw.githubusercontent.com/selva86/datasets/master/Advertising.csv'
df = pd.read_csv(url)

# Remove unnecessary index column if it exists
if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])

print(df.head())
```

## Q3. Explore the Dataset

```python
print('Dataset shape:', df.shape)
print('\nColumn names:')
print(df.columns.tolist())

print('\nDataset information:')
print(df.info())

print('\nSummary statistics:')
print(df.describe())

print('\nMissing values:')
print(df.isnull().sum())
```

## Q4. Prepare Features and Target

```python
# Independent variables
X = df[['TV', 'Radio', 'Newspaper']]

# Dependent variable
y = df['Sales']

print('Features:')
print(X.head())

print('\nTarget:')
print(y.head())
```

## Q5. Split the Data into Training and Testing Sets

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print('Training data shape:', X_train.shape)
print('Testing data shape:', X_test.shape)
```

## Q6. Build and Train the Polynomial Regression Model

We use polynomial degree 2 because it is simple and can capture curve-shaped relationships without becoming too complex.

```python
# Polynomial Regression pipeline
degree = 2

poly_model = Pipeline(steps=[
    ('polynomial_features', PolynomialFeatures(degree=degree, include_bias=False)),
    ('scaler', StandardScaler()),
    ('linear_regression', LinearRegression())
])

# Train the model
poly_model.fit(X_train, y_train)

print('Polynomial Regression model trained successfully.')
```

## Q7. Make Predictions

```python
y_pred = poly_model.predict(X_test)

results = pd.DataFrame({
    'Actual Sales': y_test.values,
    'Predicted Sales': y_pred
})

print(results.head(20))
```

## Q8. Evaluate the Model

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print('Model Evaluation Results')
print('Mean Absolute Error:', mae)
print('Mean Squared Error:', mse)
print('Root Mean Squared Error:', rmse)
print('R-squared Score:', r2)
```

## Q9. Compare Linear Regression and Polynomial Regression

```python
linear_model = Pipeline(steps=[
    ('scaler', StandardScaler()),
    ('linear_regression', LinearRegression())
])

linear_model.fit(X_train, y_train)
linear_pred = linear_model.predict(X_test)

comparison = pd.DataFrame({
    'Model': ['Linear Regression', 'Polynomial Regression'],
    'MAE': [mean_absolute_error(y_test, linear_pred), mean_absolute_error(y_test, y_pred)],
    'RMSE': [np.sqrt(mean_squared_error(y_test, linear_pred)), np.sqrt(mean_squared_error(y_test, y_pred))],
    'R2 Score': [r2_score(y_test, linear_pred), r2_score(y_test, y_pred)]
})

print(comparison)
```

## Q10. Visualize Actual vs Predicted Sales

```python
plt.figure(figsize=(8, 5))
plt.scatter(y_test, y_pred)
plt.xlabel('Actual Sales')
plt.ylabel('Predicted Sales')
plt.title('Actual vs Predicted Sales - Polynomial Regression')
plt.show()
```

## Q11. Predict Sales for Custom Advertising Budget

```python
# Example custom budget values
custom_budget = pd.DataFrame({
    'TV': [150],
    'Radio': [30],
    'Newspaper': [20]
})

custom_prediction = poly_model.predict(custom_budget)

print('Custom Advertising Budget:')
print(custom_budget)
print('\nPredicted Sales:', custom_prediction[0])
```

## Q12. Diminishing Returns Analysis

```python
# This function shows how predicted sales change when one channel budget increases
# while other channels are kept fixed at their median values.

def plot_diminishing_returns(channel):
    budget_range = np.linspace(df[channel].min(), df[channel].max(), 100)
    
    scenario = pd.DataFrame({
        'TV': np.full(100, df['TV'].median()),
        'Radio': np.full(100, df['Radio'].median()),
        'Newspaper': np.full(100, df['Newspaper'].median())
    })
    
    scenario[channel] = budget_range
    predicted_sales_curve = poly_model.predict(scenario)
    
    plt.figure(figsize=(8, 5))
    plt.plot(budget_range, predicted_sales_curve)
    plt.xlabel(f'{channel} Advertising Budget')
    plt.ylabel('Predicted Sales')
    plt.title(f'Diminishing Returns Curve for {channel}')
    plt.show()
    
    marginal_gain = np.diff(predicted_sales_curve)
    print(f'Average initial marginal gain for {channel}:', np.mean(marginal_gain[:20]))
    print(f'Average final marginal gain for {channel}:', np.mean(marginal_gain[-20:]))

plot_diminishing_returns('TV')
plot_diminishing_returns('Radio')
plot_diminishing_returns('Newspaper')
```
